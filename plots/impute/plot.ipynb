```python
# --- 1KG general imputation, restricted to admixed individuals (R1.7) ---
# Same models and reference panels as the general experiment; only the target
# set changes, to the 212 haplotypes (106 individuals) of the held-out test
# split whose 1KG population is admixed: CLM, MXL, PEL, PUR (American, admixed
# by construction) plus ASW and ACB (recently admixed African ancestry).

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import os

# --- Settings ---
num_bins = 9
bins = np.logspace(-4, np.log10(0.5), num_bins + 1)
bin_centers = np.sqrt(bins[:-1] * bins[1:])  # For plotting

all_methods = [
    'real_8020', 'pc_real_8020', 'hclt_8020', 'rbm_8020', 'wgan_8020',
    'pc_hmm_8020', 'hmm_8020',
]
all_colors = ['black', 'dodgerblue', 'b', 'orange', 'g', 'mediumorchid', 'purple']
all_markers = ['o', 's', 's', '^', 'D', 'v', 'v']
all_labels = [
    'Impute5', f'{my_method_name} (direct)', f'{my_method_name}', 'RBM', 'WGAN',
    'HMM (direct)', 'HMM',
]
all_linestyles = ['--', '-', '-', '-', '-', '-', '-']

# --- Function to summarize bootstrap statistics ---
def summarize_bootstrap(df_subset, boot_cols, label):
    if df_subset.empty:
        print(f"  {label}: No data")
        return
    bootstrap_means = df_subset[boot_cols].mean(skipna=True)
    overall_mean = bootstrap_means.mean()
    sem = bootstrap_means.std(ddof=1)
    ci_lower = overall_mean - 1.96 * sem
    ci_upper = overall_mean + 1.96 * sem
    print(f"  {label} Mean R\u00b2: {overall_mean:.4f}")
    print(f"  {label} 95% CI: [{ci_lower:.4f}, {ci_upper:.4f}]")

# --- Load data ---
method_data = {}
for method in all_methods:
    path = f'results/r2/{method}_admixed_bootstraps.csv'
    if not os.path.exists(path):
        print(f'Method: {method} -- results not available yet, skipping')
        continue
    df = pd.read_csv(path)

    # Remove variants with MAF=0 upfront
    df = df[df['MAF'] > 0].copy()

    r2_cols = [c for c in df.columns if c.startswith("R2")]
    df[r2_cols] = df[r2_cols].fillna(0)

    boot_cols = [c for c in df.columns if c.startswith('R2_boot_')]

    print(f"Method: {method}")
    summarize_bootstrap(df, boot_cols, "Full set")
    summarize_bootstrap(df[df['MAF'] < 0.01], boot_cols, "Low-freq (<1%)")
    summarize_bootstrap(df[df['MAF'] < 0.001], boot_cols, "Rare (<0.1%)")

    method_data[method] = df

# --- Plot each method with CI bands ---
plt.figure(figsize=(8,6))

for method_name, color, marker, label, linestyle in zip(
    all_methods, all_colors, all_markers, all_labels, all_linestyles
):
    if method_name not in method_data:
        continue
    df = method_data[method_name]

    boot_cols = [c for c in df.columns if c.startswith('R2_boot_')]
    bin_indices = np.digitize(df['MAF'], bins) - 1

    mean_r2, ci_low, ci_high = [], [], []
    for i in range(num_bins):
        in_bin = bin_indices == i
        if not np.any(in_bin):
            mean_r2.append(np.nan)
            ci_low.append(np.nan)
            ci_high.append(np.nan)
            continue
        boot_values = df.loc[in_bin, boot_cols].values
        bin_means = np.nanmean(boot_values, axis=0)
        mean_r2.append(np.mean(bin_means))
        ci_low.append(np.percentile(bin_means, 2.5))
        ci_high.append(np.percentile(bin_means, 97.5))

    mean_r2 = np.array(mean_r2)
    ci_low = np.array(ci_low)
    ci_high = np.array(ci_high)

    plt.plot(bin_centers, mean_r2, marker=marker, linestyle=linestyle, color=color, label=label)
    plt.fill_between(bin_centers, ci_low, ci_high, color=color, alpha=0.2)

# --- Plot formatting ---
plt.xscale('log')
plt.xticks([1e-4, 1e-3, 1e-2, 1e-1], [r"$10^{-4}$", r"$10^{-3}$", r"$10^{-2}$", r"$10^{-1}$"], fontsize=16)
plt.yticks(fontsize=16)
plt.xlabel('MAF', fontsize=18)
plt.ylabel(r'Average $r^2$', fontsize=18)
plt.grid(True, which='both', linestyle='--', linewidth=0.5)
plt.legend(fontsize=14)
plt.tight_layout()
plt.savefig("results/plots/8020_admixed_bootstrap_ci.pdf", dpi=300, bbox_inches='tight')
plt.show()
```

```python
# --- High-coverage 1KG (array-based), restricted to admixed individuals (R1.7) ---
# Same models and reference panels as the b38 general experiment; only the target
# set changes, to the 212 haplotypes (106 individuals) of the held-out split whose
# 1KG population is admixed. The b38 split uses the same 501 individuals as the
# 1KG split, so this is the identical subset as the 1KG admixed cell above.

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import os

# --- Settings ---
num_bins = 9
all_methods = ['real', 'pc', 'hclt', 'rbm', 'wgan', 'pc_hmm', 'hmm']
all_colors = ['black', 'dodgerblue', 'b', 'orange', 'g', 'mediumorchid', 'purple']
all_markers = ['o', 's', 's', '^', 'D', 'v', 'v']
all_labels = [
    'Impute5', f'{my_method_name} (direct)', f'{my_method_name}', 'RBM', 'WGAN',
    'HMM (direct)', 'HMM',
]
all_linestyles = ['--', '-', '-', '-', '-', '-', '-']

# --- Function to summarize bootstrap statistics ---
def summarize_bootstrap(df_subset, boot_cols, label):
    if df_subset.empty:
        print(f"  {label}: No data")
        return
    bootstrap_means = df_subset[boot_cols].mean(skipna=True)
    overall_mean = bootstrap_means.mean()
    sem = bootstrap_means.std(ddof=1)
    ci_lower = max(overall_mean - 1.96 * sem, 0)
    ci_upper = min(overall_mean + 1.96 * sem, 1)
    print(f"  {label} Mean R\u00b2: {overall_mean:.4f}")
    print(f"  {label} 95% CI: [{ci_lower:.4f}, {ci_upper:.4f}]")

# --- Load data ---
method_data = {}
for method in all_methods:
    path = f'results/multi/8020_multi_{method}_b38_hum5_admixed_chr15_results.csv'
    if not os.path.exists(path):
        print(f'Method: {method} -- results not available yet, skipping')
        continue
    df = pd.read_csv(path)
    df = df[df['MAF'] > 0].copy().reset_index(drop=True)

    r2_cols = [col for col in df.columns if col.startswith("R2")]
    df[r2_cols] = df[r2_cols].fillna(0)
    boot_cols = [c for c in df.columns if c.startswith('R2_boot_')]

    print(f"Method: {method}")
    summarize_bootstrap(df, boot_cols, "Full set")
    summarize_bootstrap(df[df['MAF'] < 0.01], boot_cols, "Low-freq (<1%)")
    summarize_bootstrap(df[df['MAF'] < 0.001], boot_cols, "Rare (<0.1%)")

    method_data[method] = df

# --- Equal-count MAF bins, as in the other b38 cells ---
bin_edges = np.quantile(pd.concat([d['MAF'] for d in method_data.values()]),
                        np.linspace(0, 1, num_bins + 1))
bin_centers = np.sqrt(bin_edges[:-1] * bin_edges[1:])

# --- Plot each method with CI bands ---
plt.figure(figsize=(8,6))

for method_name, color, marker, label, linestyle in zip(
    all_methods, all_colors, all_markers, all_labels, all_linestyles
):
    if method_name not in method_data:
        continue
    df = method_data[method_name]
    boot_cols = [c for c in df.columns if c.startswith('R2_boot_')]

    bin_indices = np.digitize(df['MAF'], bin_edges) - 1
    bin_indices[bin_indices == num_bins] = num_bins - 1

    mean_r2, ci_low, ci_high = [], [], []
    for i in range(num_bins):
        in_bin = bin_indices == i
        if not np.any(in_bin):
            mean_r2.append(np.nan); ci_low.append(np.nan); ci_high.append(np.nan)
            continue
        boot_values = df.loc[in_bin, boot_cols].values
        bin_means = np.nanmean(boot_values, axis=0)
        mean = np.mean(bin_means)
        sem = np.std(bin_means, ddof=1)
        mean_r2.append(mean)
        ci_low.append(max(mean - 1.96 * sem, 0))
        ci_high.append(min(mean + 1.96 * sem, 1))

    mean_r2 = np.array(mean_r2); ci_low = np.array(ci_low); ci_high = np.array(ci_high)
    plt.plot(bin_centers, mean_r2, marker=marker, linestyle=linestyle, color=color, label=label)
    plt.fill_between(bin_centers, ci_low, ci_high, color=color, alpha=0.2)

# --- Plot formatting ---
plt.xscale('log')
plt.xticks([1e-3, 1e-2, 1e-1], [r"$10^{-3}$", r"$10^{-2}$", r"$10^{-1}$"], fontsize=16)
plt.yticks(fontsize=16)
plt.ylim(0,1)
plt.xlabel('MAF', fontsize=18)
plt.ylabel(r'Average $r^2$', fontsize=18)
plt.grid(True, which='both', linestyle='--', linewidth=0.5)
plt.legend(fontsize=14)
plt.tight_layout()
plt.savefig("results/plots/b38_8020_multi_admixed_bootstrap_ci.pdf", dpi=300, bbox_inches='tight')
plt.show()
```
